## Feature Engineering

```python
def feature_engineering(data):
    data = data.copy()

    # Total area
    data['TotalSF'] = (
        data['TotalBsmtSF'].fillna(0) +
        data['1stFlrSF'].fillna(0) +
        data['2ndFlrSF'].fillna(0)
    )

    # Total bathrooms
    data['TotalBathrooms'] = (
        data['FullBath'].fillna(0) +
        0.5 * data['HalfBath'].fillna(0) +
        data['BsmtFullBath'].fillna(0) +
        0.5 * data['BsmtHalfBath'].fillna(0)
    )

    # House age
    data['HouseAge'] = data['YrSold'] - data['YearBuilt']

    # Age since remodeling
    data['RemodAge'] = data['YrSold'] - data['YearRemodAdd']

    # Total porch area
    porch_cols = [
        'OpenPorchSF',
        '3SsnPorch',
        'EnclosedPorch',
        'ScreenPorch',
        'WoodDeckSF'
    ]

    data['TotalPorchSF'] = data[porch_cols].fillna(0).sum(axis=1)

    return data


# Apply feature engineering
df_model = feature_engineering(df)

print("New Shape:", df_model.shape)
```

## Features and Target

```python
X = df_model.drop(columns=[target])
y = df_model[target]
print("X Shape:", X.shape)
print("y Shape:", y.shape)
```

## Train-Test Split

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.20, random_state=42)
print("Training data:", X_train.shape)
print("Testing data:", X_test.shape)
```

## Preprocessing Pipeline

```python
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder

numeric_features = X_train.select_dtypes(include=np.number).columns
categorical_features = X_train.select_dtypes(exclude=np.number).columns

numeric_pipeline = Pipeline([('imputer', SimpleImputer(strategy='median'))])
categorical_pipeline = Pipeline([
    ('imputer', SimpleImputer(strategy='most_frequent')),
    ('encoder', OneHotEncoder(handle_unknown='ignore', sparse_output=False))
])

preprocessor = ColumnTransformer([
    ('num', numeric_pipeline, numeric_features),
    ('cat', categorical_pipeline, categorical_features)
])
```

## Model Definition

```python
from sklearn.linear_model import LinearRegression, Ridge, Lasso
from sklearn.ensemble import RandomForestRegressor, GradientBoostingRegressor
from sklearn.tree import DecisionTreeRegressor
from xgboost import XGBRegressor

models = {
    'Linear Regression': LinearRegression(),
    'Ridge': Ridge(alpha=10),
    'Lasso': Lasso(alpha=0.001, max_iter=10000),
    'Decision Tree': DecisionTreeRegressor(random_state=42),
    'Random Forest': RandomForestRegressor(n_estimators=300, random_state=42, n_jobs=-1),
    'Gradient Boosting': GradientBoostingRegressor(random_state=42),
    'XGBoost': XGBRegressor(n_estimators=500, learning_rate=0.05, max_depth=6,
                            subsample=0.8, colsample_bytree=0.8,
                            objective='reg:squarederror', random_state=42, n_jobs=-1)
}
```

## Train and Evaluate Models

```python
from sklearn.metrics import r2_score, mean_absolute_error, mean_squared_error

results = []
trained_models = {}

for name, regressor in models.items():
    pipeline = Pipeline([('preprocessor', preprocessor), ('model', regressor)])
    pipeline.fit(X_train, y_train)
    y_pred = pipeline.predict(X_test)
    r2 = r2_score(y_test, y_pred)
    mae = mean_absolute_error(y_test, y_pred)
    mse = mean_squared_error(y_test, y_pred)
    rmse = np.sqrt(mse)
    results.append({'Model': name, 'R2': r2, 'MAE': mae, 'MSE': mse, 'RMSE': rmse})
    trained_models[name] = pipeline

results_df = pd.DataFrame(results).sort_values('R2', ascending=False)
display(results_df)
```

## Model Performance Plot

```python
plt.figure(figsize=(12, 6))
sns.barplot(data=results_df, x='R2', y='Model')
plt.title("Model Comparison Based on R² Score")
plt.xlabel("R² Score")
plt.ylabel("Model")
plt.xlim(0, 1)
plt.show()
```

## Best Model

```python
best_model_name = results_df.iloc[0]['Model']
best_model = trained_models[best_model_name]
print("Best Model:", best_model_name)
print("Best R2 Score:", results_df.iloc[0]['R2'])
```

## Actual vs Predicted

## XGBoost Hyperparameter Tuning

```python
from sklearn.model_selection import RandomizedSearchCV

xgb_pipeline = Pipeline([
    ('preprocessor', preprocessor),
    ('model', XGBRegressor(objective='reg:squarederror', random_state=42, n_jobs=-1))
])

param_grid = {
    'model__n_estimators': [300, 500, 700],
    'model__learning_rate': [0.01, 0.03, 0.05, 0.1],
    'model__max_depth': [3, 4, 5, 6, 8],
    'model__min_child_weight': [1, 3, 5],
    'model__subsample': [0.7, 0.8, 0.9, 1.0],
    'model__colsample_bytree': [0.7, 0.8, 0.9, 1.0]
}

random_search = RandomizedSearchCV(
    estimator=xgb_pipeline,
    param_distributions=param_grid,
    n_iter=30,
    cv=5,
    scoring='r2',
    random_state=42,
    n_jobs=-1,
    verbose=1
)
random_search.fit(X_train, y_train)
print("Best Parameters:")
print(random_search.best_params_)
print("Best Cross Validation R2:", random_search.best_score_)
```

## Tuned Model Evaluation

```python
tuned_model = random_search.best_estimator_
tuned_pred = tuned_model.predict(X_test)
tuned_r2 = r2_score(y_test, tuned_pred)
tuned_mae = mean_absolute_error(y_test, tuned_pred)
tuned_mse = mean_squared_error(y_test, tuned_pred)
tuned_rmse = np.sqrt(tuned_mse)
print("Tuned XGBoost Performance")
print("--------------------------------")
print("R2   :", tuned_r2)
print("MAE  :", tuned_mae)
print("MSE  :", tuned_mse)
print("RMSE :", tuned_rmse)
```

## Customer Price Prediction Function

```python
def predict_customer_house(overall_quality, living_area, basement_area,
                            garage_cars, year_built, full_bathrooms,
                            bedrooms, lot_area):
    house = pd.DataFrame({
        'OverallQual': [overall_quality],
        'GrLivArea': [living_area],
        'TotalBsmtSF': [basement_area],
        'GarageCars': [garage_cars],
        'YearBuilt': [year_built],
        'FullBath': [full_bathrooms],
        'BedroomAbvGr': [bedrooms],
        'LotArea': [lot_area]
    })
    return customer_model.predict(house)[0]

predicted_price = predict_customer_house(
    overall_quality=7, living_area=1800, basement_area=900,
    garage_cars=2, year_built=2010, full_bathrooms=2,
    bedrooms=3, lot_area=8000
)
print(f"Predicted House Price: ${predicted_price:,.2f}")
```

## Customer Recommendation System

```python
def recommend_house(budget, min_area=0, min_quality=0,
                    min_bedrooms=0, min_bathrooms=0,
                    preferred_neighborhood=None, top_n=5):

    data = df.copy()

    if 'GrLivArea' in data.columns:
        data = data[data['GrLivArea'] >= min_area]

    if 'OverallQual' in data.columns:
        data = data[data['OverallQual'] >= min_quality]

    if 'BedroomAbvGr' in data.columns:
        data = data[data['BedroomAbvGr'] >= min_bedrooms]

    if 'FullBath' in data.columns:
        data = data[data['FullBath'] >= min_bathrooms]

    if preferred_neighborhood is not None and 'Neighborhood' in data.columns:
        neighborhood_data = data[
            data['Neighborhood'].str.lower() == preferred_neighborhood.lower()
        ]
        if len(neighborhood_data) > 0:
            data = neighborhood_data

    data = data[data[target] <= budget]

    if len(data) == 0:
        return pd.DataFrame(columns=data.columns)

    data['ValueScore'] = (
        data['OverallQual'].fillna(data['OverallQual'].median()) * 2
        + data['GrLivArea'].fillna(data['GrLivArea'].median()) / 1000
        + data['FullBath'].fillna(0)
        + data['BedroomAbvGr'].fillna(0)
    )

    return data.sort_values('ValueScore', ascending=False).head(top_n)


recommendations = recommend_house(
    budget=300000,
    min_area=1500,
    min_quality=6,
    min_bedrooms=3,
    min_bathrooms=2,
    top_n=5
)

display(recommendations[[col for col in [
    'Id', 'Neighborhood', 'OverallQual', 'GrLivArea',
    'BedroomAbvGr', 'FullBath', 'YearBuilt',
    'SalePrice', 'ValueScore'
] if col in recommendations.columns]])
```
